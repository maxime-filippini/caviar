```python
import numpy as np
import matplotlib.pyplot as plt

f_tmin1 = -0.03
y_tmin1 = np.linspace(-0.05, 0.05, 150)
y_tmin1_pos = np.maximum(y_tmin1, 0)
y_tmin1_neg = np.maximum(-y_tmin1, 0)

b0_as, b1_as, b2_as, b3_as = [-0.00233754, 0.84092582, 0.01236809, -0.50435605]
f_t_as = b0_as + b1_as * f_tmin1 + b2_as * y_tmin1_pos + b3_as * y_tmin1_neg

b0_sav, b1_sav, b2_sav = [-0.00384236, 0.71771244, -0.51741154]
f_t_sav = b0_sav + b1_sav * f_tmin1 + b2_sav * np.abs(y_tmin1)

b0_ig, b1_ig, b2_ig = [7.18902662e-05, 7.37432974e-01, 1.28563631e00]
f_t_ig = -np.sqrt(b0_ig + b1_ig * f_tmin1**2 + b2_ig * y_tmin1**2)

with plt.style.context("minimalist.mplstyle"):
    fig, (ax, ax1, ax2) = plt.subplots(1, 3, figsize=(14, 4))

    ax.set_title("Indirect GARCH")
    ax.plot(y_tmin1, -f_t_ig, label=r"Indirect GARCH $\beta=(0.1, 0.85, 0.12)$")
    ax.scatter([0], [-f_tmin1], marker="x", color="red")
    ax.annotate(
        r"$f_{t-1}$", (0, -f_tmin1), xytext=(-8, 10), textcoords="offset points"
    )

    ax1.set_title("Symmetric Absolute Value")
    ax1.plot(
        y_tmin1, -f_t_sav, label=r"Symmetric Absolute Value $\beta=(-0.2, 0.85, -0.15)$"
    )
    ax1.scatter([0], [-f_tmin1], marker="x", color="red")
    ax1.annotate(
        r"$f_{t-1}$", (0, -f_tmin1), xytext=(-8, 10), textcoords="offset points"
    )

    ax2.set_title("Asymmetric Slope")
    ax2.plot(
        y_tmin1, -f_t_as, label=r"Asymmetric Slope $\beta=(-0.2, 0.8, 0.05, -0.3)$"
    )
    ax2.annotate(
        r"$f_{t-1}$", (0, -f_tmin1), xytext=(-8, 10), textcoords="offset points"
    )
    ax2.scatter([0], [-f_tmin1], marker="x", color="red")

    plt.savefig("nics.png")

    plt.show()
```
